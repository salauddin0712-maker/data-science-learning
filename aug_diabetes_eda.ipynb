```python
#Replacing 0 with NaN---

import numpy as np

cols = ['Glucose', 'BloodPressure', 'SkinThickness',
        'Insulin', 'BMI']

diabetes_df[cols] = diabetes_df[cols].replace(0 , np.nan)
```
